```python
import numpy as np
from sentence_transformers import SentenceTransformer
print("Environment is ready!!")
```

```python
# Load the embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
# encode the text
text = "I love programming"
embedding = model.encode(text)
print(embedding)
```

```python
print(type(embedding))
```

```python
print(embedding.shape)
```

```python
# let's check the first 10 numbers of the embedding list
print(embedding[:10])
```

```python
# Now let's see the same thing for more than one sentence

sentences = [
  "I love programming",
  "I enjoy writing code",
  "I like eating pizza"
]

embedding_sentences = model.encode(sentences)
print(embedding_sentences.shape)
```

* Sentence 1 → 384 numbers
* Sentence 2 → 384 numbers
* Sentence 3 → 384 numbers

#### to check the semantically similar sentences, the most common used method is **COSINE SIMILARITY**

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
scores = cosine_similarity(
    [embedding_sentences[0]],
    embedding_sentences[1:]
)
print(scores)
```

# **A small project: semantic search engine**

a working system that finds one most relevant document.

```python
# create our own document list

documents = [
    "Python is a popular programming language used for web development, automation, and data science.",
    "Machine learning allows computers to learn patterns from data.",
    "Docker is a tool used to package applications and their dependencies into containers.",
    "Python functions are reusable blocks of code that perform a specific task.",
    "SQL is a language used to communicate with and query relational databases.",
    "Git is a version control system used to track changes in source code."
]

print(len(documents))
```

```python
# convert the doc texts into numbers
docs_embeddings = model.encode(documents)

print(docs_embeddings.shape) # 6 sentences with 384 dimensions each
```

```python
# user's query
query = "How can I learn Python programming?"

# query embedding
q_embed = model.encode(query)
print(q_embed.shape)
```

```python
sim_scores = cosine_similarity(
    [q_embed],
    docs_embeddings
)

print(sim_scores)
```

```python
# check the highest similarity score

best_index = np.argmax(sim_scores)
print(best_index)
```

```python
# retrieve the document of the at the best index

print(documents[best_index])
```

# **Build Top-K semantic Search**
return the top 3 most relevant documents

```python
documents = [
  "Python is a popular programming language used for web development, automation, and data science.",
  "Machine learning allows computers to learn patterns from data.",
  "Docker is a tool used to package applications and their dependencies into containers.",
  "Python functions are reusable blocks of code that perform a specific task.",
  "SQL is a language used to communicate with and query relational databases.",
  "Git is a version control system used to track changes in source code."
]

print(len(documents))
```

```python
# convert the document text into vectors

document_embed = model.encode(documents)

print(document_embed.shape)
```

```python
# user's query

query = "How does the docker work?"

q_embedding = model.encode(query)
print(q_embedding.shape)
```

```python
similarity_scores = cosine_similarity(
    [q_embedding],
    document_embed
)

print(similarity_scores)
```

sort the scores in descending order for finding top-3

```python
ascending_index = np.argsort(similarity_scores[0])
print(ascending_index)

reversed_index = ascending_index[::-1]
print(reversed_index)
```

```python
top_3_indices = reversed_index[:3]
print(top_3_indices)
```

```python
print(documents[0])
print(documents[3])
print(documents[1])
```

```python
for i in top_3_indices:
  print(documents[i])
```

##**Final project: Mini semantic search engine**

Goal:

Build a program where a user can type any query, and your system returns the top 3 most relevant documents with their similarity scores.

```python
# list of our documents
documents = [
    "Python is a popular programming language used for web development, automation, data analysis, and artificial intelligence. Its simple syntax makes it beginner-friendly.",

    "Machine learning is a branch of artificial intelligence where computers learn patterns from data. Common applications include prediction, classification, recommendation systems, and image recognition.",

    "Retrieval-Augmented Generation, or RAG, combines information retrieval with large language models. Relevant documents are retrieved first and then provided to the language model as context for generating an answer.",

    "SQL is a language used to store, retrieve, and manipulate data in relational databases. It can be used to filter records, join tables, insert data, and update existing information.",

    "Docker is a platform for packaging applications and their dependencies into lightweight containers. Containers help developers run software consistently across different environments.",

    "Git is a distributed version control system used to track changes in source code. Developers use Git to create branches, collaborate with teammates, review changes, and maintain project history.",

    "Cloud computing provides computing resources such as servers, storage, databases, and networking over the internet. Popular cloud platforms include services for application hosting, data storage, and scalable infrastructure.",

    "An API allows different software applications to communicate with each other. REST APIs commonly use HTTP methods such as GET, POST, PUT, and DELETE to exchange data.",

    "React is a JavaScript library for building user interfaces. It allows developers to create reusable components and efficiently update the user interface when application data changes.",

    "Cybersecurity focuses on protecting computers, networks, applications, and data from unauthorized access and attacks. Important areas include authentication, authorization, encryption, and vulnerability management."
]
```

```python
# create the embeddings of our documents
embedding_docs = model.encode(documents)

# shape of the docs embeddings
print(embedding_docs.shape)
```

### Queries to test:
1. How can I learn a programming language easily?
2. What technology can I use to package my application?
3. How can software systems communicate with each other?
4. What can I use to protect my application?
5. How does RAG provide information to an AI model?
6. Where can I store and query structured data?

```python
user_query = "How can I learn a programming language easily?"

user_query_embed = model.encode(user_query)
print(user_query_embed.shape)
```

```python
# calculate the similarity scores between the user query and the documents given
sim_scores = cosine_similarity(
    [user_query_embed],
    embedding_docs
)

print(sim_scores)
```

```python
asc_index_sim_scores = np.argsort(sim_scores[0])
print(asc_index_sim_scores)
```

```python
reverse_index_sim_scores = asc_index_sim_scores[::-1]
print(reverse_index_sim_scores)
```

```python
top_3 = reverse_index_sim_scores[:3]
print(top_3)
```

```python
for i in top_3:
  print(documents[i])
```

Let's say threshold is 0.3

```python
best_score_index = top_3[0]
print(best_score_index)
```

```python
best_score = sim_scores[0][best_score_index]
print(best_score)
```

```python
if best_score < 0.3:
  print("No relevant document is found!")
else:
  for i in top_3:
    print(documents[i])
```
